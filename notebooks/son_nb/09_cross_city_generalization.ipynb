# 09 — Cross-city Generalization

Thiết kế Leave-One-City-Out để kiểm tra unseen-city generalization. Chưa dùng city_id làm feature chính.

```python
from pathlib import Path
import pandas as pd
PROJECT_ROOT=Path.cwd().parents[2]; SON_OUTPUT=PROJECT_ROOT/'notebooks'/'son_nb'/'output'; WORK_DATA=SON_OUTPUT/'data'; WORK_TABLE=SON_OUTPUT/'tables'; MULTI=WORK_DATA/'five_cities_standardized_working.csv'
```

```python
if MULTI.exists():
    df=pd.read_csv(MULTI,parse_dates=['datetime']); cities=sorted(df.city_id.dropna().unique())
    loco=pd.DataFrame([{'test_city':t,'train_cities':', '.join([c for c in cities if c!=t])} for t in cities])
    display(loco); loco.to_csv(WORK_TABLE/'loco_experiment_plan.csv',index=False)
else: print('Run Notebook 08 first.')
```

### Logic
Ví dụ: Train = Beijing + Shanghai + Guangzhou + Chengdu; Test = Shenyang.

Xoay vòng từng city. Đây là validation mạnh hơn random row split cho mục tiêu generalization. Sau LOCO China mới thử city ngoài China.
